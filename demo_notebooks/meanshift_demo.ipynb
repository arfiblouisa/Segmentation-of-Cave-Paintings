# Segmentation of cave paintings using Mean Shift algorithm

This notebook contains examples of the segmentation of cave paintings using the Mean Shift algorithm. The parameters that are used are those which have been empirically proven the best in the report : $h_s=80$ (for an image of size $340 \times 400$), and $h_r=24$.

As mentionned in the README, for the segmentation to be as accurate as possible, you need to provide the coordinates of a bounding box that encapsulates the painting you want to segment. You can find these coordinates by running the dedicated script.

```python
import seg_cave_paintings.preprocessing_utils as prep_utils
import seg_cave_paintings.meanshift as ms_utils
```

## Segmentation of an auroch from Lascaux cave (France)

### With the parametres guarantizing the best tradeoff between the number of regions to merge and the reuslts

```python
img_path='../dataset/lascaux/Lascaux_021.jpg'
x1, y1, x2, y2 = 402, 198, 923, 669

k_s=80
k_r=24

preprocessed_img=prep_utils.crop_and_preprocess_img(img_path, x1, y1, x2, y2)
unique_colors, segmented_rgb=ms_utils.create_regions_image(preprocessed_img, k_s, k_r)
```

```python
regions_to_merge=[2,3]

ms_utils.create_segmentation_results(regions_to_merge, unique_colors, segmented_rgb, preprocessed_img)
```

### With smaller bandwidth parameters: more regions to merge but better results

```python
img_path='../dataset/lascaux/Lascaux_021.jpg'
x1, y1, x2, y2 = 402, 198, 923, 669

k_s=64
k_r=16

preprocessed_img=prep_utils.crop_and_preprocess_img(img_path, x1, y1, x2, y2)
unique_colors, segmented_rgb=ms_utils.create_regions_image(preprocessed_img, k_s, k_r)
```

```python
regions_to_merge=[2,3,4,6,7]

ms_utils.create_segmentation_results(regions_to_merge, unique_colors, segmented_rgb, preprocessed_img)
```
