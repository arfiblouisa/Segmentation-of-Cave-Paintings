# Segmentation of cave paintings using Morphological Geodesic Active Contours

This notebook contains examples of the segmentation of cave paintings using morphological geodesic active contours. The parameters that are used are those which have been empirically proven the best in the report : num_iter=500 and threshold=0.75. 

As mentionned in the README, for the segmentation to be as accurate as possible, you need to provide the coordinates of a bounding box that encapsulates the painting you want to segment. You can find these coordinates by running the dedicated script.

```python
from skimage.morphology import remove_small_objects
import seg_cave_paintings.preprocessing_utils as prep_utils
import seg_cave_paintings.morphological_active_contours as mgac_utils
```

## Segmentation of an auroch from Lascaux cave (France)

```python
img_path='../dataset/lascaux/Lascaux_021.jpg'
x1, y1, x2, y2 = 402, 198, 923, 669

preprocessed_img=prep_utils.crop_and_preprocess_img(img_path, x1, y1, x2, y2)

snake_mask, evolution=mgac_utils.apply_active_contours(preprocessed_img) # Apply active contours to segment the image
mgac_utils.show_contour(preprocessed_img, snake_mask)

h,w=snake_mask.shape
snake_mask_filtered = remove_small_objects(snake_mask.astype(bool), min_size=h*w/100) # Filter out small objects (region size < 1% of total image size) from the binary mask
mgac_utils.show_contour(preprocessed_img, snake_mask_filtered, save_path=None)

mgac_utils.create_segmentation_results(snake_mask_filtered, preprocessed_img, save_path=None) # Visualize the final segmentation results
```
