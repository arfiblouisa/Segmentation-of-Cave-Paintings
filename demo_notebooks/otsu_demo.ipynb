# Segmentation of cave paintings using a method based on thresholds

The following algorithm is based on a threshold on the L channel (in L\*a\*\*b color space) and the S channel (in HSV color space).


```python
# utils

import cv2
import numpy as np
import matplotlib.pyplot as plt

import seg_cave_paintings.preprocessing_utils as prep_utils
import seg_cave_paintings.otsu as otsu_utils
```

```python
img_path="../dataset/cosquer/Panneau_des_chevaux_-_Cosquer_Mediterranee.jpg"
x1,y1,x2,y2=100, 375, 543, 682

save_path="./segmentation_results/otsu_thresholding/horse_Lascaux_segmentation_mask.png"

horse_preprocessed_rgb = prep_utils.crop_image(img_path, x1,y1,x2,y2)

mask_horse = otsu_utils.otsu_mask_from_color_space(horse_preprocessed_rgb)
mask_horse = mask_horse > 0
otsu_utils.create_segmentation_results(mask_horse, horse_preprocessed_rgb, save_path=None)
```
